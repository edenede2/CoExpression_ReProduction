```r
# Load required library
library(SummarizedExperiment)

# Extract counts matrix and transpose (donors as rows, genes as columns)
counts_matrix <- t(assay(expr, "counts"))

# Modify row names: extract the numeric part and add "ROSMAP-" prefix
rownames(counts_matrix) <- paste0("ROSMAP-", gsub("_.*", "", rownames(counts_matrix)))

# Convert to data frame and write to CSV
write.csv(counts_matrix, "ROSMAP_tissue_quad.csv", row.names = TRUE)
```

```r
# Function to process RDS file and generate CSV with normalized donor names
process_rds_to_csv <- function(rds_path, output_csv) {
    # Read the expression rds file
    expr_data <- readRDS(rds_path)
    
    # Extract counts matrix and transpose (donors as rows, genes as columns)
    counts_matrix <- t(assay(expr_data, "counts"))
    
    # Modify row names: extract the numeric part and add "ROSMAP-" prefix
    rownames(counts_matrix) <- paste0("ROSMAP-", gsub("_.*", "", rownames(counts_matrix)))
    
    # Convert to data frame and write to CSV
    write.csv(counts_matrix, output_csv, row.names = TRUE)
    
    # Return message
    cat("CSV file saved to:", output_csv, "\n")
}

# Example usage:
# process_rds_to_csv("/path/to/your/file.Rds", "output_name.csv")
```

```r
process_rds_to_csv("/home/psylab-6028/Downloads/DTS216_2024-07-11_motorTriplicates/data/sma_resids.Rds", "ROSMAP_tissue_sma.csv")
process_rds_to_csv("/home/psylab-6028/Downloads/DTS216_2024-07-11_motorTriplicates/data/vh_resids.Rds", "ROSMAP_tissue_vh.csv")
```
