```python
def dataset_details(df):
    num_clusters = df['Cluster ID'].nunique()
    num_ct_clusters = len(df[df['Cluster Type'] == 'CT'])
    num_nonct_clusters = len(df[df['Cluster Type'] == 'TS'])
    percent_ct_genes = (df[df['Cluster Type'] == 'CT'].shape[0] / df.shape[0]) * 100
    percent_nonct_genes = (df[df['Cluster Type'] == 'TS'].shape[0] / df.shape[0]) * 100
    df_ts = df[df['Cluster Type'] == 'TS']
    df_ct = df[df['Cluster Type'] == 'CT']
    AC_dom_ts = len(df_ts[df_ts['Dominant Tissue'] == 'AC'])
    PCG_dom_ts = len(df_ts[df_ts['Dominant Tissue'] == 'PCGBA23'])
    MF_dom_ts = len(df_ts[df_ts['Dominant Tissue'] == 'MFBA9BA46'])
    percent_AC_dom_ts = (AC_dom_ts / df_ts.shape[0]) * 100
    percent_PCG_dom_ts = (PCG_dom_ts / df_ts.shape[0]) * 100
    percent_MF_dom_ts = (MF_dom_ts / df_ts.shape[0]) * 100
    AC_dom_ct = len(df_ct[df_ct['Dominant Tissue'] == 'AC'])
    PCG_dom_ct = len(df_ct[df_ct['Dominant Tissue'] == 'PCGBA23'])
    MF_dom_ct = len(df_ct[df_ct['Dominant Tissue'] == 'MFBA9BA46'])
    percent_AC_dom_ct = (AC_dom_ct / df_ct.shape[0]) * 100
    percent_PCG_dom_ct = (PCG_dom_ct / df_ct.shape[0]) * 100
    percent_MF_dom_ct = (MF_dom_ct / df_ct.shape[0]) * 100
    details = {
        'num_clusters': num_clusters,
        'num_ct_clusters': num_ct_clusters,
        'num_nonct_clusters': num_nonct_clusters,
        'percent_ct_genes': percent_ct_genes,
        'percent_nonct_genes': percent_nonct_genes,
        'amount_AC_dom_ts': AC_dom_ts,
        'percent_AC_dom_ts': percent_AC_dom_ts,
        'amount_PCG_dom_ts': PCG_dom_ts,
        'percent_PCG_dom_ts': percent_PCG_dom_ts,
        'amount_MF_dom_ts': MF_dom_ts,
        'percent_MF_dom_ts': percent_MF_dom_ts,
        'percent_AC_dom_ct': percent_AC_dom_ct,
        'percent_PCG_dom_ct': percent_PCG_dom_ct,
        'percent_MF_dom_ct': percent_MF_dom_ct
    }
    return details
    
```

```python
datasets = {'df1': df1, 'df2': df2, 'df3': df3}
all_details = {}
for name, dataset in datasets.items():
    all_details[name] = dataset_details(dataset)
all_details
```
