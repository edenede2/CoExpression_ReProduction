## Generate full network of GTEx

Here we going to generate the full network of olds and youngs for 3 tissues, brain, muscle and adipose.

- Load and concate the expression files.

```r
old_adipose = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Adipose - Subcutaneous_old.csv"
old_brain = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Brain - Cortex_old.csv"
old_muscle = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Muscle - Skeletal_old.csv"
young_adipose = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Adipose - Subcutaneous_young.csv"
young_brain = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Brain - Cortex_young.csv"
young_muscle = r"/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/GTEx_outputs/Muscle - Skeletal_young.csv"
```

```r
import pandas as pd 

adipose = pd.concat([pd.read_csv(old_adipose), pd.read_csv(young_adipose)], axis=0)
brain = pd.concat([pd.read_csv(old_brain), pd.read_csv(young_brain)], axis=0)
muscle = pd.concat([pd.read_csv(old_muscle), pd.read_csv(young_muscle)], axis=0)
```

- save the full csvs

```r
import os

if not os.path.exists("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks"):
    os.makedirs("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks")

adipose.to_csv("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/adipose_full.csv", index=False)
brain.to_csv("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/brain_full.csv", index=False)
muscle.to_csv("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/muscle_full.csv", index=False)
```

- load the wgcna module builder scripts

```r
source("/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/old_scripts/ClusteringBuilding.R")
```

- set the tissues names and files paths

```r
gtex_combined_tissues_files <-c(
    "/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/adipose_full.csv",
    "/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/brain_full.csv",
    "/media/psylab-6028/DATA/Eden/CoExpression_ReProduction/nbs/GTEx_full_networks/muscle_full.csv"
)

tissue_names <- c("adipose", "brain", "muscle")
```

- run the wgcna to get the modules

```r
XWGCNA_Clusters_autoBeta(
    tissue_names = tissue_names,
    tissue_expr_file_names = gtex_combined_tissues_files,
    sd_quantile = 0.0,
    max_genes_per_tissue = 20000,     
    out_prefix = "xwgcna_gtex_full_network",
    auto_beta = TRUE,
    plot_beta_curves = TRUE,
    blockwise_TOM = TRUE,
    TS_power = 3,
    CT_power = 2,
    group = "xwgcna_gtex_full_network",
    beta_method = "wgcna_new",
    save_intermediates = TRUE,
    wgcna_powerVector = seq(0.5, 20, length.out = 20),
    ct_fisher = FALSE,
    ct_fisher_scheme = "to_ref",
    ct_fisher_Nref = "max",
    ct_too_few_action = "zero",
    adj_type = "new"
)
```
