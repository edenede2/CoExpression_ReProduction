```python
import pandas as pd

def clean_data(df):
    # Filter rows based on column: 'Cluster.ID'
    df = df[df['Cluster.ID'] == 110]
    # Sort by column: 'GeneRatio' (descending)
    df = df.sort_values(['GeneRatio'], ascending=[False])
    return df

# Loaded variable 'df' from URI: /Volumes/Transcend/ROSMAP_WGCNA/kegg_rosmap_brains_se2_res3.csv
df = pd.read_csv(r'/Volumes/Transcend/ROSMAP_WGCNA/kegg_rosmap_brains_se2_res3.csv',on_bad_lines='skip')

df_clean = clean_data(df.copy())
df_clean.head()
```
